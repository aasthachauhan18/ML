```python
Structure of a Decision Tree Node

Each box in the tree (node) usually contains something like this:

petal length (cm) <= 2.45
gini = 0.0
samples = 50
value = [50, 0, 0]
class = setosa


Here’s what each line means:

1. petal length (cm) <= 2.45

This is the decision rule for splitting.

If True, you go to the left child; if False, you go to the right child.

2. gini = 0.0
This is the Gini impurity, a measure of how mixed the classes are in that node.

0.0 means the node is pure (all samples belong to one class).

Higher values (max ~0.5 for 2 classes) mean more mixture.

3.samples = 50

Number of samples (data points) in that node.

4.value = [50, 0, 0]

Distribution of the samples among the classes.

In this example, [50, 0, 0] means 50 samples of class 0 (setosa), 0 of class 1, 0 of class 2.

5.class = setosa

The predicted class for this node.

If a sample reaches this node, the tree predicts setosa.
```

```python
Suppose a node says:
petal width (cm) <= 1.75
gini = 0.5
samples = 54
value = [0, 28, 26]
class = versicolor

If your sample’s petal width ≤ 1.75, you go left.

This node is impure (Gini = 0.5), so it contains a mix of 28 versicolor and 26 virginica.

The tree predicts versicolor because it’s the majority class here.
```
